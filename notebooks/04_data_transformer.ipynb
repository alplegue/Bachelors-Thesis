# CSV Result Transformer

Utility to parse and consolidate raw PSO result CSV files.

```python
import re
import csv


INPUT_FILE = 'resultados_pso2.csv'
OUTPUT_FILE = 'results.csv'

pat_iter     = re.compile(r'(?i)\bIteration\s+(\d+)\s*[:|]')
pat_cost     = re.compile(r'Cost\(Fitness\)\s*=\s*([\d\.Ee+-]+)')
pat_time     = re.compile(r'Time:\s*([\d\.]+)')
pat_diversity = re.compile(r'Diversity:\s*([\d\.]+)')
pat_vector   = re.compile(r'\[([^\]]*)\]')

run_number = 0
rows = []

with open(INPUT_FILE, 'r', encoding='utf-8') as f:
    for line in f:
        line = line.strip()
        if not line:
            continue

        m_iter = pat_iter.search(line)
        if m_iter:
            iteration = int(m_iter.group(1))

            # Iteration 0 signals a new run
            if iteration == 0:
                run_number += 1

            # Extract fitness
            m_cost = pat_cost.search(line)
            cost = m_cost.group(1) if m_cost else ''

            # Extract time
            m_time = pat_time.search(line)
            time_val = m_time.group(1) if m_time else ''

            # Extract position vector (optional)
            m_vec = pat_vector.search(line)
            position_vec = m_vec.group(1).strip() if m_vec else ''

            rows.append([run_number, iteration, cost, time_val, position_vec])

# Write CSV
with open(OUTPUT_FILE, 'w', newline='', encoding='utf-8') as csv_file:
    writer = csv.writer(csv_file)
    writer.writerow(['run', 'iteration', 'cost_fitness', 'iteration_time', 'best_global_position'])
    for row in rows:
        writer.writerow(row)
```
